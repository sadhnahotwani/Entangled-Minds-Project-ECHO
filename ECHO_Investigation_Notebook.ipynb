## Module 01: The Gatekeeper (Mandatory)
* **Bronze:** Produce a bit always 1, and a bit 50/50 uncertain.
* **Silver:** Two-gate sequence resulting in deterministic outcome ($H \rightarrow H$).
* **Gold:** Produce exact 3-bit pattern `101` from `000`.

```python
# Bronze & Gold Implementation for Module 01
simulator = AerSimulator()

# Gold Vault Code: 101 from 000
qc_m1_gold = QuantumCircuit(3, 3)
qc_m1_gold.x(0)  # Qubit 0 -> 1
qc_m1_gold.x(2)  # Qubit 2 -> 1 (Qubit 1 remains 0)
qc_m1_gold.measure([0, 1, 2], [0, 1, 2])

job = simulator.run(transpile(qc_m1_gold, simulator), shots=1000)
result = job.result()
counts = result.get_counts(qc_m1_gold)
print("Module 01 Gold Result (101):", counts)
```

## Module 03: The Two Observers (Mandatory)
* **Bronze:** Build twin detectors that always agree (`00` or `11`).
* **Silver/Gold:** Bell state correlation and basis-switching verification.

```python
# Module 03: Bell State Correlation
qc_m3 = QuantumCircuit(2, 2)
qc_m3.h(0)
qc_m3.cx(0, 1)
qc_m3.measure([0, 1], [0, 1])

job_m3 = simulator.run(transpile(qc_m3, simulator), shots=1000)
counts_m3 = job_m3.result().get_counts(qc_m3)
print("Module 03 Detector Agreement Counts:", counts_m3)
```

## Module 05: The Damaged Message (Mandatory)
* Recover original 32-bit message from 96-bit corrupted stream via majority voting.

```python
# Corrupted stream: 96 bits as 32 triples
corrupted_stream = [
    "100", "011", "001", "001", "001", "011", "100", "110", 
    "001", "011", "000", "000", "001", "100", "011", "011", 
    "000", "111", "000", "000", "110", "000", "001", "000", 
    "000", "111", "001", "010", "101", "011", "111", "111"
]

recovered_bits = []
for triple in corrupted_stream:
    # Majority voting rule: if 2 or more '1's, bit is '1', else '0'
    bit = '1' if triple.count('1') >= 2 else '0'
    recovered_bits.append(bit)

recovered_binary = "".join(recovered_bits)
print("Recovered Binary Message:", recovered_binary)
```

## Module 06: The Impossible Claim (Stretch Module)
* Run teleportation circuit and verify classical channel dependence.

```python
# Module 06: Quantum Teleportation Circuit
qc_m6 = QuantumCircuit(3, 3)

# 1. Prepare state to teleport on qubit 0
qc_m6.h(0)

# 2. Create EPR pair between qubit 1 and 2
qc_m6.h(1)
qc_m6.cx(1, 2)

# 3. Bell measurement on sender side (qubit 0 and 1)
qc_m6.cx(0, 1)
qc_m6.h(0)
qc_m6.measure([0, 1], [0, 1])

# 4. Receiver conditional corrections
qc_m6.x(2).c_if(1, 1)
qc_m6.z(2).c_if(0, 1)
qc_m6.measure(2, 2)

job_m6 = simulator.run(transpile(qc_m6, simulator), shots=1000)
counts_m6 = job_m6.result().get_counts(qc_m6)
print("Module 06 Teleportation Results:", counts_m6)
```
